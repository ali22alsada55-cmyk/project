```python
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import matplotlib.pyplot as plt

# قراءة الملف من المجلد الذي أنشأته على جوجل درايف
df = pd.read_csv('/content/drive/My Drive/cybersecurity_project_data/security_log.csv')

# عرض البيانات
print("بيانات سجلات الأمان:")
print(df)

# رسم بياني توضيحي
df.plot(kind='bar', x='status', y='count', color=['green', 'red', 'orange'], legend=False)
plt.title('Security Log Analysis - Cybersecurity Project')
plt.xlabel('Status Type')
plt.ylabel('Count')
plt.savefig('/content/drive/My Drive/cybersecurity_project_data/analysis_result.png')
plt.show()
```
